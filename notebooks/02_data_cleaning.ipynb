```python
import pandas as pd
import numpy as np

df = pd.read_csv("../dataset/raw/Telco-Customer-Churn.csv")
```

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
)

df["TotalCharges"] = df["TotalCharges"].fillna(0)
```

```python
df.to_csv(
    "../dataset/processed/cleaned_telco.csv",
    index=False
)
```
