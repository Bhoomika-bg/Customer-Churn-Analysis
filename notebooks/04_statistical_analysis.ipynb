## Categorical Variable Effect Sizes (Cramér's V)

**Cramér's V** extends the Chi-square test by reporting the strength of association between two categorical variables. It ranges from 0 (no association) to 1 (perfect association). Unlike a p-value, Cramér's V is not directly affected by sample size.

The formula is:

`V = sqrt(chi-square / (n * (k - 1)))`

where `n` is the number of observations and `k` is the smaller of the row or column category counts.

```python
from scipy.stats import chi2_contingency

categorical_chi_square_variables = [
    "Contract", "InternetService", "OnlineSecurity",
    "PaymentMethod", "Dependents", "Partner"
]

chi_square_results = []
for variable in categorical_chi_square_variables:
    contingency_table = pd.crosstab(df[variable], df["Churn"])
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    n = contingency_table.to_numpy().sum()
    min_dimension = min(contingency_table.shape)
    cramers_v = np.sqrt(chi2 / (n * (min_dimension - 1)))
    chi_square_results.append({
        "Variable": variable,
        "Test": "Chi-square",
        "Statistic": chi2,
        "p-value": p_value,
        "Degrees of Freedom": dof,
        "Cramer's V": cramers_v,
        "Decision": "Reject H0" if p_value < 0.05 else "Fail to reject H0"
    })

pd.DataFrame(chi_square_results).sort_values("Cramer's V", ascending=False)
```

## Mann–Whitney U Test: Tenure vs Churn

**Why this test?** `tenure` is numerical and `Churn` creates two independent groups. The Mann–Whitney U test compares distributions without requiring normally distributed tenure data.

- **Null hypothesis (H0):** The tenure distributions of churned and retained customers are equal.
- **Alternative hypothesis (H1):** The tenure distributions differ between the two groups.
- **Significance level:** alpha = 0.05.
- **Decision rule:** Reject H0 when p-value < 0.05.

```python
tenure_stayed = df.loc[df["Churn"] == "No", "tenure"]
tenure_churned = df.loc[df["Churn"] == "Yes", "tenure"]

tenure_statistic, tenure_p_value = mannwhitneyu(
    tenure_churned,
    tenure_stayed,
    alternative="two-sided"
)

print("Tenure median for retained customers:", tenure_stayed.median())
print("Tenure median for churned customers:", tenure_churned.median())
print("Mann-Whitney U statistic:", tenure_statistic)
print("p-value:", tenure_p_value)
print("Decision:", "Reject H0" if tenure_p_value < 0.05 else "Fail to reject H0")
```

## Mann–Whitney U Test: MonthlyCharges vs Churn

**Why this test?** `MonthlyCharges` is numerical and churned and retained customers form two independent groups. The test evaluates whether the two distributions differ.

- **Null hypothesis (H0):** The monthly-charge distributions are equal for both churn outcomes.
- **Alternative hypothesis (H1):** The monthly-charge distributions differ by churn outcome.
- **Significance level:** alpha = 0.05.

```python
charges_stayed = df.loc[df["Churn"] == "No", "MonthlyCharges"]
charges_churned = df.loc[df["Churn"] == "Yes", "MonthlyCharges"]

charges_statistic, charges_p_value = mannwhitneyu(
    charges_churned,
    charges_stayed,
    alternative="two-sided"
)

print("MonthlyCharges median for retained customers:", charges_stayed.median())
print("MonthlyCharges median for churned customers:", charges_churned.median())
print("Mann-Whitney U statistic:", charges_statistic)
print("p-value:", charges_p_value)
print("Decision:", "Reject H0" if charges_p_value < 0.05 else "Fail to reject H0")
```
